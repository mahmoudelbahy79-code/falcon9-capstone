# 01 · Data Collection with SpaceX API Snapshot

This notebook inspects the historical API-response snapshot distributed with IBM Skills Network course materials. It is a frozen educational snapshot, not a live API pull.

```python
from pathlib import Path
import json, pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
api_path = ROOT / "data" / "API_call_spacex_api.json"
launches = json.loads(api_path.read_text())
print(f"Launch records in snapshot: {len(launches)}")
print("Fields:", sorted(launches[0].keys()))
pd.json_normalize(launches).head(5)
```
