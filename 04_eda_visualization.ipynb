# 04 · EDA with Data Visualization

Generate the annual trend, site comparisons, orbit rates, and payload comparison from the 90-row labeled table. Rates are always interpreted alongside sample sizes.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA = ROOT / "data"
RESULTS = ROOT / "results"
import matplotlib.pyplot as plt
df = pd.read_csv(DATA / "falcon9_labeled.csv", parse_dates=["Date"])
annual = df.groupby(df.Date.dt.year).Class.agg(["size","mean"])
site = df.groupby("LaunchSite").Class.agg(["size","sum","mean"]).sort_values("mean")
orbit = df.groupby("Orbit").Class.agg(["size","sum","mean"]).sort_values("size",ascending=False)
annual, site, orbit.head(10)
```

```python
fig,ax=plt.subplots(figsize=(9,4)); annual["mean"].mul(100).plot(kind="bar",ax=ax,color="#00bcd4")
ax.set(title="Annual positive landing-label rate",xlabel="Launch year",ylabel="Rate (%)"); ax.set_ylim(0,100); ax.grid(axis="y",alpha=.2); fig.tight_layout(); plt.show()
```

```python
fig,ax=plt.subplots(figsize=(9,4)); (100*site["mean"]).plot(kind="barh",ax=ax,color="#00bcd4")
ax.set(title="Landing-label rate by launch site",xlabel="Rate (%)",ylabel="Launch site"); ax.set_xlim(0,100); fig.tight_layout(); plt.show()
```

```python
fig,ax=plt.subplots(figsize=(7,4)); df.boxplot(column="PayloadMass",by="Class",ax=ax)
ax.set(title="Payload mass by target class",xlabel="Class (0=other, 1=positive landing label)",ylabel="Payload mass (kg)"); fig.suptitle(""); plt.tight_layout(); plt.show()
```
