# 03 · Data Wrangling

The capstone modeling file contains 90 labeled Falcon 9 launches. Missing `LandingPad` values are excluded from the feature set; the target remains the course-provided `Class` label.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA = ROOT / "data"
RESULTS = ROOT / "results"
df = pd.read_csv(DATA / "dataset_part_2.csv", parse_dates=["Date"])
print(df.shape, df.Date.min(), df.Date.max())
print(df["Class"].value_counts().sort_index())
print(df.isna().sum().sort_values(ascending=False).head())
```

```python
# Predictor set excludes identifiers and fields that can leak the outcome.
num = ["PayloadMass", "Flights", "Block", "ReusedCount"]
cat = ["Orbit", "LaunchSite", "GridFins", "Reused", "Legs"]
X = df[num + cat]
y = df["Class"]
print("Numerical features:", num)
print("Categorical features:", cat)
print("Target counts:", y.value_counts().to_dict())
```

`Class=0` combines all other coded outcomes, including some missions without a landing attempt or with unspecified outcome; it does not represent failure conditional on an attempt.
