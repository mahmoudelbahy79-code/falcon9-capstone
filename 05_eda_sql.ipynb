# 05 · EDA with SQL

Load the curated dataset into SQLite and execute the project aggregation queries.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA = ROOT / "data"
RESULTS = ROOT / "results"
import sqlite3
df = pd.read_csv(DATA / "falcon9_labeled.csv")
con = sqlite3.connect(":memory:")
df.to_sql("spacex", con, index=False, if_exists="replace")
```

```python
pd.read_sql_query("""SELECT LaunchSite, COUNT(*) AS launches, SUM(\"Class\") AS successes, ROUND(100.0*AVG(\"Class\"),1) AS rate_pct FROM spacex GROUP BY LaunchSite ORDER BY rate_pct DESC""", con)
```

```python
pd.read_sql_query("""SELECT CAST(substr(Date,1,4) AS INTEGER) AS year, COUNT(*) AS launches, SUM(\"Class\") AS successes, ROUND(100.0*AVG(\"Class\"),1) AS rate_pct FROM spacex GROUP BY year ORDER BY year""", con)
```

```python
pd.read_sql_query("""SELECT Orbit, COUNT(*) AS n, SUM(\"Class\") AS successes, ROUND(100.0*AVG(\"Class\"),1) AS rate_pct FROM spacex GROUP BY Orbit ORDER BY n DESC""", con)
```
